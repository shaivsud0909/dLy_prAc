```python
import pandas as pd
from db import get_connection

conn = get_connection()

# Read read-only (CSV-backed) Lakehouse table
df = pd.read_sql(
    "SELECT * FROM dbo.data",
    conn
)

conn.close()

print(df.head())
print(df.shape)
```

```python
df.columns = [
    "age", "sex", "cp", "trestbps", "chol", "fbs",
    "restecg", "thalach", "exang", "oldpeak",
    "slope", "ca", "thal", "num"
]
```

```python
df.head()
```

```python
# Replace '?' with NaN
df = df.replace("?", None)

# Convert columns to numeric
for col in df.columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.head()
```

```python
df = df.iloc[1:].reset_index(drop=True)

```
